# BAB II
# PEMBAHASAN

## 2.1 Dasar Teori
Praktikum pertama ini berfokus pada pengenalan konsep-konsep dasar sintaks dalam bahasa pemrograman Python. Berikut adalah poin-poin teori dasar yang mendasari pelaksanaan praktikum ini:
1. Fungsi Output (print): Fungsi bawaan Python yang digunakan untuk menampilkan teks, angka, atau nilai variabel ke media keluaran (konsol/terminal).
2. Komentar (Comments): Teks dalam kode program yang diabaikan oleh penerjemah Python (interpreter). Komentar digunakan untuk dokumentasi. Tanda """ (triple quote) dapat digunakan untuk membuat komentar beberapa baris (multi-line comment).
3. Penggabungan String (String Concatenation): Proses menyambung dua atau lebih tipe data string menjadi satu menggunakan operator tambah (+).
4. Karakter Escape (Escape Character): Karakter khusus yang diawali dengan tanda garis miring terbalik (\) untuk menyisipkan instruksi format tertentu. Contohnya \n untuk membuat baris baru (newline) dan \\ untuk menampilkan karakter garis miring terbalik.

## 2.2 Kode Program

### 2.2.1 Program "program1.py"

```python
# Program pertama
"""
Komentar
dua baris atau lebih
pertemuan 1 praktikum alpr
~~~
"""
print("Hello World")
print()
print("Program ini menghasilkan")
print("Empat baris output")
print("/ \\ // \\\\ /// \\\\\\")
print("Program ini" + " dibuat oleh a\n" + "Haykal Muhammad Dahnil 2511532014")
```

**Penjelasan :**
1. Baris 1-7: Terdapat komentar satu baris (# Program pertama) yang menggunakan (#) dan komentar banyak baris (""" ... """) yang berfungsi sebagai dokumnetasi program. Bagian ini tidak akan dieksekusi
2. Baris 8-11: Kode print("Hello World), print("Program ini menghasilkan") dan print("Empat baris output") yang berfungsi menampilkan teks sesuai dengan yang ada didalama tanda petik dua (""), sedangkan kode print() tanpa argumen berfungsi mencetak baris kosong
3. Baris 12: Menampilkan karakter simbol garis miring. Karakter gand \\ digunakan sebagai escape character agar satu karakter \ dapat tercetak dengan benar pada layar
4. Baris 13: Menggabungkana tiga potongan karakter string menggunakan operator (+) serta menysipkan karaketr \n untuk memindahkan teks did epannya ke baris baru
