## 2.1 Dasar Teori
Praktikum pertama ini berfokus pada operator pada python. Berikut adalah poin-poin teori dasar yang mendasari pelaksanaan praktikum ini:
1. Percabangan if dan if-else: Percabangan if mengeksekusi blok kode hanya jika kondisi bernilai True. Percabangan if-else menyediakan jalur alternatif ketika kondisi utama bernilai False.
2. Percabangan if-elif-else: Digunakan untuk mengevaluasi banyak kondisi secara berurutan. Begitu salah satu kondisi bernilai True, blok kode terkait dieksekusi dan seluruh kondisi di bawahnya diabaikan (mutual exclusion).
3. Percabangan Multi-IF: Menggunakan beberapa instruksi if terpisah secara sejajar. Setiap if dievaluasi secara independen sehingga memungkinkan beberapa blok kode dieksekusi sekaligus jika kondisinya sama-sama terpenuhi (contoh: akumulasi diskon).
4. Struktur match-case: Bentuk percabangan yang mencocokkan nilai variabel terhadap pola/kasus tertentu (case). Tanda case _: digunakan sebagai default case ketika tidak ada pola yang cocok.

### 2.2.4 Program "if_elif_else1_2511532014.py"

```python
umur_2014 = int(input("Input umur anda: "))
sim_2014 = input("Apakah anda sudah punya SIM C (y/t): ")[0]

if umur_2014 >= 17 and sim_2014 == 'y':
    print("Anda sudah dewasa dan sudah boleh bawa motor")
elif umur_2014 >= 17 and sim_2014 != 'y':
    print("Anda sudah dewasa tetapi tidak boleh bawa motor")
elif umur_2014 < 17 and sim_2014 == 'y':
    print("Anda belum cukup umur untuk punya SIM")
else:
    print("Anda belum cukup umur untuk bawa motor")
print("Program Selesai")
```

3. Baris 12: Mencetak penanda program selesai

### 2.2.5 Program "multif_if2_2511532014.py"

```python
#input dari user
total_belanja_2014 = float(input("Masukkan total belanja (Rp): "))

#input status member (mengecek apakah user mengetik y atau ya)
input_member_2014 = input("Apakah anda member? (y/t): ").strip().lower()
is_member_2014 = input_member_2014 in ["y", "ya"]

# input status kode promo
input_promo_2014 = input("Apakah kode promo valid? (y/t): ").strip().lower()
kode_promo_valid_2014 = input_promo_2014 in ["y", "ya"]

total_diskon_persen_2014 = 0

# multi if terpisah: setiap kondisi diperiksa secara independen
# diskon bisa ditumpuk (akumulasi) jika memenuhi beberapa syarat sekaligus

if total_belanja_2014 > 1000000:
    total_diskon_persen_2014 += 10

if is_member_2014:
    total_diskon_persen_2014 += 5

if kode_promo_valid_2014:
    total_diskon_persen_2014 += 15

#  menghitung nominal diskon dan total bayar
nominal_diskon_2014 = total_belanja_2014 * (total_diskon_persen_2014 / 100)
total_bayar_2014 = total_belanja_2014 - nominal_diskon_2014


# output hasil
print("\n--- Rincian Pembayaran ---")
print(f"Total Diskon  : {total_diskon_persen_2014}% (Rp {nominal_diskon_2014:,.0f})")
print(f"Total Bayar   : Rp {total_bayar_2014:,.0f}")

print(f"Total diskon yang anda dapatkan : {total_diskon_persen_2014}%")
# output: Total diskon yang anda dapatkan: 30% jika belanja > 1jt, member dan kode promo valis
```

**Penjelasan :**
1. Baris 2–11: Mengambil data belanja serta status member dan promo. lalu mengevaluasi ke tipe boolena menggunakan operator keanggotaan in
2. Baris 18-24: Memanfaatkan sifat multi_if untuk akumulasi penumpukan persentantase diskon (total_diskokn_persen_2014 += ...). jika ketiga syarat terpenuhi. diskon akakn dijumlahkan seluruhnya hingga mencapai 30%
3. Baris 27-34: menghitung nnominal diskondan total pembayaran lalu mencetak rincian pembayaran dengan format f-string

### 2.2.6 Program "match_case1_2511532014.py"

```python
bulan_2014 = int(input("Masukkan angka bulat (1-12): "))

match bulan_2014:
    case 1:
        print("Januari")
    case 2:
        print("Februari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Angka tidak valid")
```

**Penjelasan :**
1. Baris 1: Mengambil input angka bulat rentang 1-12 dari pengguna
2. Baris 3-29: Menggunakan satruktur match-case untuk mencocokkan nlai variabel bulan_2014 dengan nama bulan yang sesuai. Kasus case_: digunakan sebagai penanganan jika masukan di luar rentang 1-12

```python

print("=== SISTEM LOKET ALPRO ADVENTURE PARK ===")

# Input Data Pengunjung
nama_pengunjung_2014 = input("Masukkan Nama Pengunjung        : ")
umur_2014 = int(input("Input umur anda                 : "))
sim_2014 = input("Apakah Anda Sudah Punya SIM C (y/t): ").strip().lower()[0]

print("\nPilihan Paket Wahana (1-5):")
print("1. Safari Rimba         (Rp 50,000)")
print("2. Arung Jeram          (Rp 75,000)")
print("3. Motor ATV Ekstrim    (Rp 120,000)")
print("4. Roller Coaster Kilat (Rp 100,000)")
print("5. All-Access VIP       (Rp 220,000)")

paket_2014 = int(input("Masukkan nomor paket (1-5) : "))
jumlah_tiket_2014 = int(input("Masukkan jumlah tiket : "))

if jumlah_tiket_2014 <= 0:
    print("Peringatan: Kuota tiket tidak valid!")

# Pemilihan Wahana
match paket_2014:
    case 1:
        nama_wahana_2014 = "Wahana Safari Rimba"
        harga_satuan_2014 = 50000
    case 2:
        nama_wahana_2014 = "Wahana Arung Jeram"
        harga_satuan_2014 = 75000
    case 3:
        nama_wahana_2014 = "Wahana Motor ATV Ekstrim"
        harga_satuan_2014 = 120000
    case 4:
        nama_wahana_2014 = "Wahana Roller Coaster Kilat"
        harga_satuan_2014 = 100000
    case 5:
        nama_wahana_2014 = "Wahana All-Access VIP"
        harga_satuan_2014 = 220000
    case _:
        print("Paket wahana tidak valid!")
        exit()

# Validasi izin
print("\n--- KELAYAKAN PENGENDARA WAHANA ---")

if paket_2014 == 3:
    if umur_2014 >= 17 and sim_2014 == 'y':
        print("Status Akses: Anda sudah dewasa dan boleh mengendarai ATV sendiri.")
    elif umur_2014 >= 17 and sim_2014 != 'y':
        print("Status Akses: Anda sudah dewasa tetapi tidak boleh bawa motor ATV (wajib didampingi instruktur).")
    elif umur_2014 < 17 and sim_2014 == 'y':
        print("Status Akses: Identitas tidak valid: Belum cukup umur memiliki SIM.")
    else:
        print("Status Akses: Anda belum cukup umur dan tidak boleh bawa motor ATV.")
else:
    if umur_2014 >= 10:
        print("Status Akses: Anda memenuhi batas usia untuk wahana ini.")
    else:
        print("Status Akses: Anda belum cukup umur untuk wahana ini.")

#diskon
is_member_2014 = input("Apakah Anda member? (y/t) : ").strip().lower()
kode_promo_valid_2014 = input("Apakah kode promo valid? (y/t) : ").strip().lower()

#
subtotal_2014 = harga_satuan_2014 * jumlah_tiket_2014

#  akumulasi diskon
total_diskon_persen_2014 = 0

if subtotal_2014 >= 200000:
    total_diskon_persen_2014 += 10

if is_member_2014 in ['y', 'ya']:
    total_diskon_persen_2014 += 5

if kode_promo_valid_2014 in ['y', 'ya']:
    total_diskon_persen_2014 += 15

if jumlah_tiket_2014 >= 5:
    total_diskon_persen_2014 += 5

#pembayaran
nominal_diskon_2014 = subtotal_2014 * (total_diskon_persen_2014 / 100)
total_bayar_2014 = subtotal_2014 - nominal_diskon_2014

print("\n--- AUDIT TRANSAKSI ---")
print(f"Nama Pengunjung : {nama_pengunjung_2014}")
print(f"Wahana          : {nama_wahana_2014}")
print(f"Jumlah Tiket    : {jumlah_tiket_2014}")

print("\n--- RINCIAN PEMBAYARAN ---")
print(f"Subtotal Belanja : Rp {subtotal_2014:,.0f}")
print(
    f"Total Diskon     : {total_diskon_persen_2014}% "
    f"(Rp {nominal_diskon_2014:,.0f})"
)
print(f"Total Bayar      : Rp {total_bayar_2014:,.0f}")

# bonus
if total_bayar_2014 > 300000:
    print("Catatan Layanan  : Selamat! Anda berhak mendapatkan Souvenir Gratis.")
else:
    print("Catatan Layanan  : Terima kasih telah berkunjung.")

print("\nProgram Selesai")
```
