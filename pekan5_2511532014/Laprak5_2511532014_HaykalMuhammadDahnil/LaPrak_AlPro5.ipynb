## 2.1 Dasar Teori
Praktikum ini berfokus pada operator pada python. Berikut adalah poin-poin teori dasar yang mendasari pelaksanaan praktikum ini:
1. Perulangan for dan Fungsi range(): Perulangan for digunakan untuk melakukan iterasi atas urutan nilai. Fungsi range(start, stop, step) menghasilkan deret angka bulat mulai dari start hingga sebelum stop dengan selisih step.
2. Akumulator dalam Perulangan: Variabel penampung yang nilainya terus diperbarui atau dijumlahkan pada setiap iterasi untuk menghitung total akumulasi nilai.
3. Perulangan Bertingkat (Nested Loop): Struktur perulangan di mana sebuah instruksi for berada di dalam blok for lainnya. Perulangan luar (outer loop) mengontrol baris, sedangkan perulangan dalam (inner loop) mengontrol kolom atau pencetakan karakter horizontal.
4. Pembentukan Pola Geometri: Pola karakter terbentuk dari hubungan matematis antara indeks baris perulangan luar dengan jumlah spasi dan karakter pada perulangan dalam.

## 2.2 Kode Program

### 2.2.1 Program "perulangan_for1_2511532014.py"

```python
ulang_2014 = int(input("Masukkan Jumlah perulangana: "))

for i_2014 in range(ulang_2014):
    print(f"Perulangan ke-{i_2014}")
```

**Penjelasan :**
1. Baris 1: Meminta input jumlah iterasi dari pengguna dan mengonversinya ke tipe int.
2. Baris 3–4: Menjalankan perulangan for dengan range(ulang_2014) yang mencetak indeks perulangan mulai dari $0$ hingga ulang_2014 - 1 menggunakan format f-string.


### 2.2.2 Program "perulangan_for2_2511532014.py"

```python
ulang_2014 = int(input("Masukkan jumlah perulangan: "))
print("perulangan ke-0 sampai_2014 ke-", ulang_2014-1)
for i_2014 in range(ulang_2014):
    print(i_2014, end =" ")
print()
print("perulangan ke-1 sampai_2014 ke-", ulang_2014)
for i_2014 in range(1, ulang_2014+1):
    print(i_2014, end =" ")
```

**Penjelasan :**
1. Baris 1–5: Menjalankan perulangan berbasis indeks $0$ hingga ulang_2014 - 1 dengan argumen end=" " agar angka dicetak secara horizontal ke samping.
2. Baris 6–9: Menjalankan perulangan berbasis indeks $1$ hingga ulang_2014 dengan mendefinisikan batas range(1, ulang_2014 + 1).


### 2.2.3 Program "perulangan_for3_2511532014.py"

```python
ulang_2014 = int(input("Masukkan jumlah perulangan: "))

jumlah_2014 = 0
for i_2014 in range(1, ulang_2014 + 1):
    print(i_2014, end=" ")
    jumlah_2014 = jumlah_2014 + 1

    if i_2014 < ulang_2014:
        print(" + ", end="")
    else:
        print(" = ", jumlah_2014, end="")
print()
print("Jumlah = ", jumlah_2014)
```

**Penjelasan :**
1. Baris 1–4: Menginisialisasi variabel akumulator jumlah_2014 = 0 lalu melakukan iterasi dari 1 hingga nilai batas.
2. Baris 5–10: Mengakumulasikan pencacah dan mencetak simbol operator + di antara angka, serta mencetak simbol = dan hasil total saat iterasi mencapai angka terakhir.

### 2.2.4 Program "nested_for1_2511532014.py"

```python
batas_2014 = int(input("Masukkan nilai batas: "))
for line_2014 in range(1, batas_2014 + 1):
    for j_2014 in range(1, (-1* line_2014 + batas_2014) + 1):
        print(".", end="")
    print(line_2014)
```

**Penjelasan :**
1. Baris 2–3: Perulangan luar (line_2014) mengontrol nomor baris. Perulangan dalam (j_2014) mencetak karakter titik (.) dengan jumlah berkurang pada tiap baris sesuai formula (-1 * line_2014 + batas_2014) + 1.
2. Baris 5: Mencetak angka nomor baris di ujung karakter titik pada setiap barisnya.

### 2.2.5 Program "nested_for2_2511532014.py"

```python
batas_2014 = int(input("Masukkan nilai batas: "))
for i_2014 in range(1, batas_2014 + 1):
    for j_2014 in range(1, batas_2014 + 1):
        print("*", end="")
    print()
```

**Penjelasan :**
1. Baris 2–5: Menggunakan dua perulangan bertingkat sejajar range(1, batas_2014 + 1) untuk membentuk pola persegi bintang ukuran $N \times N$.

### 2.2.6 Program "nested_for3_2511532014.py"

```python
batas_2014 = int(input("Masukkan nilai batas: "))
for i_2014 in range(batas_2014 + 1):
    for j_2014 in range(batas_2014 + 1):
        print(i_2014 + j_2014, end=" ")
    print()
```

**Penjelasan :**
1. Baris 2–5: Menghasilkan matriks penjumlahan dua dimensi di mana nilai tiap elemen kolom merupakan hasil penjumlahan dari indeks baris i_2014 dan indeks kolom j_2014.

### 2.2.7 Program "nested_for4_2511532014.py"

```python
tinggi_2014 = int(input("Masukkan tinggi pola (bilangan genap): "))

if tinggi_2014 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_2014 = tinggi_2014
    c_2014 = a_2014
    lebar_2014 = (2 * tinggi_2014) - 2

    for i_2014 in range(1, tinggi_2014 + 1):
        b_2014 = c_2014 + 1

        for j_2014 in range(1, lebar_2014 + 1):

            if i_2014 == 1 or i_2014 == tinggi_2014:
                if j_2014 == 1 or j_2014 == lebar_2014:
                    print("#", end="")
                else:
                    print("=", end="") 
            else:
                if j_2014 == 1 or j_2014 == lebar_2014:
                    print("|", end="")
                else:
                    if j_2014 == c_2014:
                        print("<", end="")
                    elif j_2014 == b_2014:
                        print(">", end="")
                    elif j_2014 == (lebar_2014 - c_2014):
                        print("<", end="")
                    elif j_2014 == (lebar_2014 - c_2014 + 1):
                        print(">", end="")
                    elif j_2014 > b_2014 and j_2014 < (lebar_2014 - c_2014):
                        print(".", end="")
                    else:
                        print(" ", end="")

        print()

        a_2014 -= 2

        if a_2014 <= 0:
            c_2014 = (-a_2014) + 2
        else:
            c_2014 = a_2014
```

**Penjelasan :**
1. Baris 3–9: Memeriksa syarat masukan tinggi bernilai genap. Menghitung lebar pola bingkai berdasarkan rumus (2 * tinggi_2014) - 2.
2. Baris 11–43: Menggunakan perulangan bertingkat untuk menggambar bingkai tepi (#, =, |) serta memosisikan simbol intan <> dan titik . secara dinamis sesuai pergeseran posisi variabel c_2014.

### 2.2.8 Program "tugasPrak5_2511532014.py"

```python
tinggi_2014 = int(input("Masukkan tinggi segitga: "))

for i_2014 in range(1, tinggi_2014 + 1):
    for j_2014 in range(tinggi_2014 - i_2014):
        print(" ", end="")
    for k_2014 in range(i_2014):
        print("*", end=" ")
    print()
```

**Penjelasan :**
1. Baris 3–8: Membentuk pola segitiga sama sisi berisi bintang (*). Perulangan pertama (j_2014) mengatur cetakan spasi agar rata tengah, dan perulangan kedua (k_2014) mencetak simbol bintang berspasi sebanyak nilai baris i_2014.

## Tugas Praktikum

**Soal :**

Studi Kasus: Pola Jam Pasir Kristal Palindromik Berbingkai (Framed Palindromic Crystal Hourglass)
Mahasiswa diminta merancang program konsol interaktif untuk mencetak pola bangun geometri dua dimensi berupa Jam Pasir Kristal Palindromik yang terbungkus rapi di dalam bingkai pembatas berpagar. Berbeda dengan pola dasar, program ini menuntut pembentukan deret angka yang bergerak dinamis menurun dan menaik secara palindromik pada setiap baris serta diselaraskan dengan poros kristal di pusat bangun.
Program menerima masukan nilai integer positif N dari pengguna yang merepresentasikan ukuran skala jam pasir (skala angka tertinggi pada puncak dan alas).
Struktur Pembentukan Pola (3 Fase + 2 Bingkai Horizontal):
1. Bingkai Pembatas Horizontal (Border Atas dan Bawah):
  ○ Karakter pagar sudut (#) di ujung kiri dan kanan.
  ○ Karakter garis ganda (=) di bagian tengah sebanyak (4 * N + 5) karakter.
2. Fase 1: Jam Pasir Atas (Reduksi Angka Menurun: Baris N turun s.d. 1):
  ○ Sisi kiri dibatasi garis tegak (|) dan satu spasi padding.
  ○ Spasi penyeimbang kiri sebanyak 2 * (N - baris).
  ○ Deret angka mundur dari baris turun ke 1, dipisahkan oleh spasi (contoh baris 4: "4 3 2 1 ").
  ○ Poros kristal tengah: karakter "<*>".
  ○ Deret angka maju dari 1 naik ke baris, diawali spasi (contoh baris 4: " 1 2 3 4").
  ○ Spasi penyeimbang kanan sebanyak 2 * (N - baris).
  ○ Satu spasi padding dan garis tegak (|) di sisi kanan.
3. Fase 2: Poros Titik Pusat Jam Pasir (Titik Nol / Singularity):
  ○ Karakter garis tegak pembatas (|) di kiri.
  ○ Spasi penyeimbang kiri sebanyak (2 * N + 1).
  ○ Karakter poros kristal tunggal "<*>".
  ○ Spasi penyeimbang kanan sebanyak (2 * N + 1).
  ○ Karakter garis tegak pembatas (|) di kanan.
4. Fase 3: Jam Pasir Bawah (Ekspansi Angka Menaik: Baris 1 naik s.d. N):
  ○ Sisi kiri dibatasi garis tegak (|) dan satu spasi padding.
  ○ Spasi penyeimbang kiri sebanyak 2 * (N - baris).
  ○ Deret angka mundur dari baris turun ke 1, dipisahkan oleh spasi.
  ○ Poros kristal tengah: karakter "<*>".
  ○ Deret angka maju dari 1 naik ke baris, diawali spasi.
  ○ Spasi penyeimbang kanan sebanyak 2 * (N - baris).
  ○ Satu spasi padding dan garis tegak (|) di sisi kanan.

Ketentuan Teknis Program

Aturan Penamaan Variabel: Setiap variabel yang dideklarasikan wajib diakhiri dengan 4 digit terakhir NIM mahasiswa (contoh: n_1234, baris_1234, spasi_1234, angka_1234). Penulisan variabel harus mematuhi kaidah case-sensitive dan bebas reserved keywords.
Masukan Dinamis: Meminta masukan ukuran N via input() dan dikonversi ke tipe data integer menggunakan int().
Struktur Nested Loop Murni: Dilarang keras menggunakan jalan pintas manipulasi string perkalian seperti print("=" * n), print(" " * spasi), atau perkalian list. Seluruh pembentukan border, deret spasi, dan deret angka WAJIB diiterasi menggunakan perulangan for bersarang dengan parameter end="".
Pindah Baris: Gunakan perintah print() tanpa argumen pada akhir setiap baris pola.
C. Contoh Alur Eksekusi Program (I/O)
=== PROGRAM JAM PASIR KRISTAL PALINDROMIK (PEKAN 5) ===
Masukkan ukuran skala jam pasir (N): 4
Uji Coba 2 (Ukuran N = 3):
=== PROGRAM JAM PASIR KRISTAL PALINDROMIK (PEKAN 5) ===
Masukkan ukuran skala jam pasir (N): 3
#=================#
| 3 2 1 <*> 1 2 3 |
|   2 1 <*> 1 2   |
|     1 <*> 1     |
|       <*>       |
|     1 <*> 1     |
|   2 1 <*> 1 2   |
| 3 2 1 <*> 1 2 3 |
#=================#


**Kode Jawaban :**

a. Kode Program "Tugas5_2511532014"

```python

print("=== PROGRAM JAM PASIR KRISTAL PALINDROMIK (PEKAN 5) ===")

n_1234 = int(input("Masukkan ukuran skala jam pasir (N): "))

for pagar_1234 in range(1):
    print("#", end="")
    for sama_1234 in range(4 * n_1234 + 5):
        print("=", end="")
    print("#")

for baris_1234 in range(n_1234, 0, -1):
    print("|", end="")
    print(" ", end="")

    for spasi_1234 in range(2 * (n_1234 - baris_1234)):
        print(" ", end="")

    for angka_1234 in range(baris_1234, 0, -1):
        print(angka_1234, end=" ")
    
    print("<*>", end="")

    for angka_1234 in range(1, baris_1234 + 1):
        print(" ", end="")
        print(angka_1234, end="")

    for spasi_1234 in range(2 * (n_1234 - baris_1234)):
        print(" ", end="")

    print(" |")

print("|", end="")
for spasi_1234 in range(2 * n_1234 + 1):
    print(" ", end="")
print("<*>", end="")
for spasi_1234 in range(2 * n_1234 + 1):
    print(" ", end="")
print("|")

for baris_1234 in range(1, n_1234 + 1):
    print("|", end="")
    print(" ", end="")

    for spasi_1234 in range(2 * (n_1234 - baris_1234)):
        print(" ", end="")

    for angka_1234 in range(baris_1234, 0, -1):
        print(angka_1234, end=" ")

    print("<*>", end="")

    for angka_1234 in range(1, baris_1234 + 1):
        print(" ", end="")
        print(angka_1234, end="")

    for spasi_1234 in range(2 * (n_1234 - baris_1234)):
        print(" ", end="")

    print(" |")

for pagar_1234 in range(1):
    print("#", end="")
    for sama_1234 in range(4 * n_1234 + 5):
        print("=", end="")
    print("#")
```

**Penjelasan :**
1. Baris 5–9 & 58–62: Mencetak garis pembatas atas dan bawah pola menggunakan kombinasi karakter # dan =.
2. Baris 11–31: Membentuk bagian kerucut atas jam pasir. Menggunakan perulangan menurun (range(n_1234, 0, -1)) untuk mencetak spasi luar, deret angka menurun, simpul pusat <*>, dan deret angka menaik secara palindromik.
3. Baris 33–39: Mencetak bagian simpul tengah jam pasir secara presisi.
4. Baris 41–56: Membentuk bagian kerucut bawah jam pasir menggunakan perulangan menaik (range(1, n_1234 + 1)) dengan susunan komponen yang simetris terhadap bagian atas.
