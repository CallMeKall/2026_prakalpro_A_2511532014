## 2.1 Dasar Teori
Berikut adalah poin-poin teori dasar yang mendasari pelaksanaan praktikum ini:
1. Tipe Data Numerik: Jenis tipe data yang digunakan untuk menyimpan nilai angka, meliputi int (bilangan bulat), float (bilangan desimal), dan complex (bilangan imajiner/kompleks).
2. Tipe Data String: Sekumpulan karakter yang diapit oleh tanda petik tunggal, ganda, atau tiga lapis ("""...""") untuk teks multibaris (multi-line string).
3. Tipe Data Boolean: Tipe data yang hanya memiliki dua nilai logika, yaitu True atau False, yang umumnya dihasilkan dari evaluasi kondisi atau operasi perbandingan.
4. Konstanta (Final): Variabel yang nilainya bersifat tetap dan tidak boleh diubah setelah diinisialisasi. Pada Python, penandaan konstanta dapat diimpor melalui from typing import Final.
5. Input dan Type Casting: Fungsi input() digunakan untuk menerima masukan dari pengguna yang secara default bernilai tipe data string. Fungsi casting seperti int() atau float() digunakan untuk mengubah tipe string tersebut menjadi angka agar dapat dihitung secara matematis.

## 2.2 Kode Program

### 2.2.1 Program "numerik_2511532014.py"

```python
#Create  a variable with integer values
a_2014 = 100
print("The type of variable having value", a_2014, "is", type(a_2014))

#Create a variable with float values
b_2014 = 10.2345
print("The type of variable having value", b_2014, "is", type(b_2014))

#Create a variable with complex values
c_2014 = 100 + 3j
print("The type of variable having value", c_2014, "is", type(c_2014))
```

### 2.2.3 Program "boolean_2511532014.py"

```python
is_Lulus_2014 = True
is_cumlaude = True

# menggunakan boolean
nilai_2014 = 85
batas_lulus_2014 = 75

#menenutkan nilain boolean darin kondisi
status_kelulusan_2014 = nilai_2014 >= batas_lulus_2014 #hasilnya akan true

print("=== Check Kelulusan ===")
print("Nilai: ", nilai_2014)
print("Apakah siswa lulus?:", status_kelulusan_2014)
if is_Lulus_2014 and is_cumlaude:
    print("Selamat anda lulus dengan predikat Cum Laude!")
```

### 2.2.4 Program "input_2511532014.py"

```python
#Program untuk menghitung luas persegi panjang
#Program ini menggunakan fungsi input() untuk menerima input dari pengguna
panjang_2014 = input("Masukkan panjang persegi panjang: ")
lebar_2014 = input("Masukkan lebar persegi panjang: ")
luas_2014 = int(panjang_2014) * int(lebar_2014)
print("Luas =", luas_2014)
```

### 2.2.5 Program "konstanta_2511532014.py"

```python
from typing import Final
PI_2014: Final = 3.14
print("pi: %f" % (PI_2014))
jari_2014 = float(input("Masukan jari-jari lingkaran: "))
luas_2014 = PI_2014 * jari_2014 * jari_2014
print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_2014, luas_2014))
```

**Penjelasan :**
1. Baris 1–3: Mengimpor modul Final dari typing untuk mendefinisikan konstanta PI_2014 sebesar 3.14, kemudian mencetaknya ke layar menggunakan string formatting %f.
2. Baris 4: Mengambil masukan nilai jari-jari lingkaran dari pengguna yang langsung dikonversi ke tipe data float menggunakan float().
3. Baris 5–6: Menghitung luas lingkaran berdasarkan rumus $L = \pi \times r \times r$ lalu mencetak hasilnya dengan format dua angka di belakang koma (%.2f).

Ketentuan Teknis Program
● Aturan Penamaan Variabel: Setiap variabel yang dibuat wajib diakhiri dengan 4 digit
terakhir NIM mahasiswa (contoh untuk NIM akhiran 1234: nama_1234, umur_1234,
nilai_1234). Penulisan variabel harus mematuhi kaidah case-sensitive dan tidak boleh
menggunakan kata kunci bawaan Python (reserved keywords).
● Penerapan Konstanta: Buat ketetapan batas kelulusan (misalnya skor 75.0)

menggunakan modul typing.Final. Konstanta dideklarasikan di awal program dengan
format penulisan huruf kapital.
● Tipe Data String & Char:
○ Ambil data nama praktikan menggunakan fungsi input().
○ Gunakan tanda petik tunggal atau ganda untuk karakter inisial / jenis kelamin.
○ Gunakan tanda petik tiga ("""...""") untuk menyimpan alamat tempat tinggal secara
multiline (lebih dari satu baris).
● Tipe Data Numerik & Type Casting:
○ Gunakan fungsi input() yang dikonversi ke tipe data Integer (int()) untuk masukan
umur atau semester.
○ Gunakan fungsi input() yang dikonversi ke tipe data Float (float()) untuk masukan
nilai ujian / skor tes awal praktikum.
○ Deklarasikan satu variabel bilangan kompleks (Complex) sebagai token identifikasi
praktikan (contoh: 100 + 3j).
● Tipe Data Boolean: Evaluasi status kelulusan praktikan secara dinamis menggunakan
operator logika perbandingan (misalnya: nilai_1234 >= BATAS_LULUS) sehingga
menghasilkan nilai True atau False.
● Pengecekan Tipe Data (type()): Program wajib menampilkan semua data yang telah
diinput dan diproses ke layar, disertai dengan tipe datanya masing-masing menggunakan
fungsi bawaan type().

**Kode Jawaban :**

a. Kode Program "Tugas2_2511532014"

```python
from typing import Final

BATAS_LULUS: Final = 75.0

# Input
nama_2014 = input("Masukkan Nama Mahasiswa : ")
jenis_kelamin_2014 = input("Masukkan Jenis Kelamin (L/P): ")
umur_2014 = int(input("Masukkan Umur : "))
nilai_2014 = float(input("Masukkan Skor Tes Awal : "))

alamat_2014 = """Pasa Ambacang,
Pauh,
Kota Padang"""

token_2014 = 100 + 3j

lulus_2014 = nilai_2014 >= BATAS_LULUS

# Output
print("\n=== DATA PRAKTIKAN & HASIL PEMERIKSAAN ===")
print("Nama Mahasiswa :", nama_2014, "| Tipe:", type(nama_2014))
print("Jenis Kelamin :", jenis_kelamin_2014, "| Tipe:", type(jenis_kelamin_2014))
print("Alamat Domisili:")
print(alamat_2014, "| Tipe:", type(alamat_2014))
print("Umur :", umur_2014, "tahun | Tipe:", type(umur_2014))
print("Skor Tes Awal :", nilai_2014, "| Tipe:", type(nilai_2014))
print("ID Token Sinyal:", token_2014, "| Tipe:", type(token_2014))

print("\n=== STATUS KELULUSAN PRAKTIKUM ===")
print("Batas Minimum Nilai:", BATAS_LULUS)
print("Apakah Dinyatakan Lulus?:", lulus_2014, "| Tipe:", type(lulus_2014))
```

**Penjelasan :**
1. Baris 1–3: Mengimpor modul Final dan mendefinisikan konstanta BATAS_LULUS sebesar 75.0 sebagai acuan nilai minimum kelulusan.
2. Baris 5–9: Mengambil masukan data dari pengguna untuk nama, jenis kelamin, umur (dikonversi ke int), serta nilai tes awal (dikonversi ke float).
3. Baris 11–15: Mendeklarasikan alamat domisili menggunakan format string multibaris ("""...""") dan ID token berbentuk bilangan kompleks (100 + 3j).
4. Baris 17: Mengevaluasi kondisi kelulusan dengan menguji apakah nilai_2014 >= BATAS_LULUS yang menghasilkan nilai bertipe boolean.
5. Baris 19–27: Mencetak seluruh data praktikan beserta jenis tipe datanya masing-masing menggunakan fungsi type().
6. Baris 29–31: Menampilkan batas acuan nilai minimum dan status akhir kelulusan praktikan beserta tipe data dari hasil evaluasi boolean tersebut.
